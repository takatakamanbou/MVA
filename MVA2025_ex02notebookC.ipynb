以下では，CSVファイルを読み込む → pandas の DataFrame というものをつくる → そこから必要なデータを抽出する，という手順で NumPy array を作っています．


```python
# pandas のパッケージをインポートして，pd という名前で呼べるようにする
import pandas as pd

# URL を指定して CSV ファイルを読み込み，DataFrame に変換．日本語が含まれているので encoding オプションを指定
df = pd.read_csv('https://www.data.jma.go.jp/cpdinfo/temp/list/csv/an_wld.csv', encoding='cp932')
df
```

```python
# df の「年」列を xvec に，「世界全体」列を yvec に代入
xvec = df['年'].to_numpy()
yvec = df['世界全体'].to_numpy()
print(xvec)
print(yvec)
```

```python
# データの散布図を描く
fig, ax = plt.subplots()
ax.scatter(xvec, yvec, s=8)
ax.set_xlim(1890, 2025)
ax.set_ylim(-1.5, 0.5)
plt.show()
```
