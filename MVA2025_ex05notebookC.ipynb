## 分散共分散行列
---

この先の授業で何度も登場する **分散共分散行列** について説明します（注）．

<hr width="50%" align="left">
<span style="font-size: 75%">
※注： 平均に関して，確率変数の分布に基づく「母平均」と，データ（標本）から計算される「標本平均」があるのと同様に，分散共分散行列についても，「母分散共分散行列」と「標本分散共分散行列」があります．ここでは，標本分散共分散行列について説明しています．
</span>


$N$ 個の $D$ 次元ベクトル $\pmb{x}_1, \pmb{x}_2, \ldots, \pmb{x}_N$ から成るデータを考えます．$\pmb{x}_n = (x_{n, 1}, x_{n, 2}, \ldots, x_{n, D})$ とします（$n = 1, 2, \ldots, N$）．

このデータの $d$ 番目の要素の平均 $\bar{x}_d$ と分散 $s_{d}^{2}$ を次のようにおきます（$d = 1, 2, ..., D$）:

$$
\begin{aligned}
\bar{x}_d &= \frac{1}{N} \sum_{n=1}^{N} x_{n, d}\\
s_d^2 &= \frac{1}{N} \sum_{n=1}^{N} (x_{n, d} - \bar{x}_d)^2
\end{aligned}
$$


<img width="400" src="https://www-tlab.math.ryukoku.ac.jp/~takataka/course/Data/ex05-correl2.png" align="right">

さらに，$i$ 番目の要素と $j$ 番目の要素の間の **共分散** $s_{i, j}$ を次のように定義します：

$$
s_{i, j} = \frac{1}{N} \sum_{n=1}^{N} (x_{n, i} - \bar{x}_i)(x_{n, j} - \bar{x}_j)
$$


共分散は，2つの変数が共に変動するその変動の仕方を表す量です．一方が平均より大きい/小さいときに他方も平均より大きい/小さいことが多ければ（正の相関を持てば）正の値をとり，一方が平均より大きい/小さいときに他方は平均より小さい/大きいことが多ければ（負の相関を持てば）負の値をとります．両者が無関係に変動するならば（相関がなければ） $0$ に近くなります（注）．

右図では $x_{n, i}, x_{n, j}$ の代わりに $X, Y$ という記号を使っていますが，この例では，横軸の値が平均より大きい/小さいときに縦軸の値は平均より小さい/大きいことが多く，両者の共分散は負となっています．

<hr width="50%" align="left">
<span style="font-size: 75%">
※注： 「データ分析」を履修したひとは，「相関係数」の説明の際に似たような話が出てきたのを覚えているでしょうか．相関係数は，共分散をそれぞれの変数の標準偏差で割ったものです．</br>
参考: <a href="https://www-tlab.math.ryukoku.ac.jp/wiki/?Data/2025#ex05">2025年度「データ分析」第5回</a>
</span>


**分散共分散行列** は，共分散を要素に持つ行列です．上記のデータの場合，分散共分散行列は，$i, j$ 要素が $s_{i, j}$ であるような $D$ 次正方行列となります．$i = j$ のとき $s_{i, j} = s_{i}^2$ ですので，その対角要素は，各変数の分散です．また，共分散の定義から $s_{i, j} = s_{j, i}$ となりますので，分散共分散行列は対称行列です（分散共分散行列を $S$ とおくと，$S^{\top} = S$）．

上では要素ごとの共分散を用いて分散共分散行列を定義しましたが，個々のデータが列ベクトルである場合，その平均を

$$
\bar{\pmb{x}} = \frac{1}{N}\sum_{n=1}^{N}\pmb{x}_n
$$

として，分散共分散行列 $S$ を次のように定義することもできます．

$$
S = \frac{1}{N}\sum_{n=1}^{N} (\pmb{x}_n - \bar{\pmb{x}})(\pmb{x}_n - \bar{\pmb{x}})^{\top}
$$

試しに，「数学，物理，情報の点数100人分のデータ」の分散共分散行列を求めてみましょう．まずはデータを読み込んで散布図を描いてみます．

```python
# データを入手
! wget -nc https://www-tlab.math.ryukoku.ac.jp/~takataka/course/MVA/mpiS100.csv
# データを読み込んで表示
dfMPI = pd.read_csv('mpiS100.csv')
#dfMPI[['数学', '物理', '情報']]

# 数学物理情報の変数間の相関
fig, ax = plt.subplots(1, 3, figsize=(9, 3))

ax[0].scatter(dfMPI['数学'], dfMPI['物理'], s=8)
ax[0].set_title(f'M vs P (1st vs 2nd)')

ax[1].scatter(dfMPI['数学'], dfMPI['情報'], s=8)
r = np.corrcoef(dfMPI['数学'], dfMPI['情報'])[0, 1]
ax[1].set_title(f'M vs I (1st vs 3rd)')

ax[2].scatter(dfMPI['物理'], dfMPI['情報'], s=8)
r = np.corrcoef(dfMPI['物理'], dfMPI['情報'])[0, 1]
ax[2].set_title(f'P vs I (2nd vs 3rd)')

for ax_ in ax:
    ax_.set_xlim(0, 100)
    ax_.set_ylim(0, 100)
    ax_.set_aspect('equal')
plt.show()
```

第1の変数 M と第2の変数 P の間の相関は小さく，M と第3の変数 I との間には負の相関があり，P と I の間には小さな正の相関があるようです．

それでは，分散共分散行列を計算してみましょう．

```python
# データのならんだ 2次元 numpy array を作る
X = dfMPI[['数学', '物理', '情報']].to_numpy()
print(f"X.shape = {X.shape}")
# 平均と分散
print(f"平均: {np.mean(X, axis=0)}")
print(f"分散: {np.var(X, axis=0)}")
print()
# 分散共分散行列を求める
X2 = X - np.mean(X, axis=0)
S = X2.T @ X2 / X2.shape[0] # 上の説明では個々のデータは列ベクトルだがここでは行ベクトルなので，転置の付く位置が逆
print("S = ")
print(S)
```

これを見ると，上で説明した性質が成り立っていることが分かります．
- 対称行列である
- 対角要素がそれぞれの変数の分散に一致している
- 第1の変数 M と第3の変数 I との間に負の相関がある = 負の共分散を持つので，$(1,3)$ 要素および $(3, 1)$ 要素が負で絶対値の大きな値になっている．$(1, 2), (2, 1), (2, 3), (3, 2)$ は，相関が小さいことに対応して絶対値が小さめ．

## 次回に備えて実験しよう
----

次回扱うデータ分析手法について，事前に少しだけ学んでおきましょう．

$N$ 個の $D$ 次元列ベクトル $\pmb{x}_1, \pmb{x}_2, \ldots, \pmb{x}_N$ に対して，$D$ 次元列ベクトル $\pmb{a}$ を定めると，

$$
y_n = \pmb{a}^{\top}\pmb{x}_n \quad (n = 1, 2, \ldots, N)
$$

という計算によって，$D$ 次元ベクトルたちをそれぞれ実数値に変換することができます（notebookAの「「行列」＝「ベクトルを並べたもの」」および「直交変換と直交行列」で解説していることです）．



ここでは，$D = 2$ の場合について，実際のデータをこのように変換する実験を行ってみましょう．



```python
# データを読み込む
dfRawData = pd.read_csv('https://www-tlab.math.ryukoku.ac.jp/~takataka/course/MVA/HighSchoolPhysicalFitnessMeasurement.csv')
dfData = dfRawData[['身長[cm]', '体重[kg]']]
dfData -= dfData.mean() # 平均を差し引く => 平均が (0, 0) になる
dfData
```

```python
# X という numpy.array に格納する．0列目が身長，1列目が体重
X = dfData.to_numpy()
print(X)
print(X.shape)
```

それでは，`X` に格納された2次元データたちを，2次元ベクトル $\pmb{a}$ で変換する実験を行いましょう．ただし，$\|\pmb{a}\| = 1$ の場合に限定して考えます（そうする理由は次回分かります）．この条件のもとでは，任意のベクトル $\pmb{a}$ を

$$
\pmb{a} = \begin{pmatrix}
\cos\theta \\
\sin\theta
\end{pmatrix} \quad ( 0 \leq \theta < 2\pi)
$$

と表せます．
次のコードセルの `theta:` と書かれた箱に数値を入力して実行すると，その数を $\theta$ として$\pmb{a}$ を求め，次のことを行います．

1. $\pmb{x}_n$ の散布図を描き，そこに $\pmb{a}$ の向きが分かるような半直線を描く（左図）．
1. $y_n = \pmb{a}^{\top}\pmb{x}_n$ を求めて，そのヒストグラムを描く．$y_n$ の分散も求める（右図）．

なお，$\pi \leq \theta < 2\pi$ のときは，$\pmb{a} = \begin{pmatrix}
-\cos (\pi-\theta) \\
-\sin (\pi -\theta)
\end{pmatrix} = - \begin{pmatrix}
\cos (\pi-\theta) \\
\sin (\pi -\theta)
\end{pmatrix}$ と表せるので，$\pi - \theta$ のときの $y_n$ の符号を反転させた結果が得られるだけです．そのため，次のセルの実験では，$0 \leq \theta < \pi$ の範囲のみを考えればokです．`theta:` の箱には数値しか書けませんので，例えば $\theta$ を $\frac{\pi}{4}$ に近い値に設定したければ `0.785` と入力する等してね．

```python
# @title 実験
theta = 0 # @param {type:"number"}

a = np.array([np.cos(theta), np.sin(theta)])
print(f'theta = {theta:.3f}, a = ({a[0]:.3f}, {a[1]:.3f})')

# X を構成する2次元ベクトルたちを a でひとつの数値に変換
y = X @ a

# y の分散
s2 = np.var(y)

# 図
fig, ax = plt.subplots(1, 2, figsize=(8, 4))
xmin, xmax = -50, 50
ymin, ymax = -50, 50

# 散布図
ax[0].scatter(X[:, 0], X[:, 1], marker='.', s=8)
ax[0].set_xlim(xmin, xmax)
ax[0].set_ylim(ymin, ymax)
ax[0].axhline(0, color='gray')
ax[0].axvline(0, color='gray')
ax[0].set_xlabel(r'$x_1$')
ax[0].set_ylabel(r'$x_2$')
ax[0].set_aspect('equal')
ax[0].plot([0, 2*xmax*a[0]], [0, 2*xmax*a[1]], color='orange', linewidth=3, label='a')
ax[0].legend()

# ヒストグラム
ax[1].hist(y, bins=20, label=f'var(y) = {s2:.1f}')
ax[1].set_xlim(xmin, xmax)
ax[1].set_xlabel(r'$y$')
ax[1].legend()

fig.tight_layout()
plt.show()
```
