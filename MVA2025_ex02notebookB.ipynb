### ［補足］ 2次元配列と1次元配列の間の演算

np.array の2次元配列は行列に，1次元配列はベクトルに相当するとみなせることが多いですが，それではつじつまの合わない場合があります（注）．ひっかかりやすいところなので，いくつか簡単に示しておきます．

<br>
<hr width="50%" align="left">
<span style="font-size: 75%">
※ 注: np.array の演算に備わる「ブロードキャスト」という仕組みが影響しています．これは，np.array を用いた様々な演算を簡単に書けるようにしてくれるものなのですが，初心者がはまりやすいところです．
</span>

次のコードセルを実行すると，一部は結果が表示されますが，一部はエラーになります．

````python
# 2次元配列と1次元配列の加算
B = np.array([[1, 2, 3], [4, 5, 6]])
print("B =")
print(B, B.shape)
x = np.array([10, 20, 30])
print("x =", x, x.shape)
y = np.array([100, 200])
print("y =", y, y.shape)
print()

# B + x
print("B + x =")
Bpx = B + x
print(Bpx, Bpx.shape)
print()
# B + y
print("B + y =")
Bpy = B + y
print(Bpy, Bpy.shape)
````

数学的には，`B + x` や `B + y` のベクトルを行ベクトルだとみなしても列ベクトルだとみなしても，これらの加算はできません．

$$
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} + \begin{pmatrix} 10 & 20 & 30\end{pmatrix} \rightarrow \text{NG} \qquad
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} + \begin{pmatrix} 10 \\ 20 \\ 30\end{pmatrix} \rightarrow \text{NG} \\
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} + \begin{pmatrix} 100 & 200 \end{pmatrix} \rightarrow \text{NG} \qquad
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} + \begin{pmatrix} 100 \\ 200\end{pmatrix} \rightarrow \text{NG} \\
$$

これにならうと，`B + x` もエラーになるべきでしょう．しかし，上記を見ると分かるように，`B + x` に関しては，`B` の列数と `x` の要素数が 3 で等しいため，NumPy が勝手に `x` を複数行にコピーしたものを作り，次のような計算を行っています．

$$
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} + \begin{pmatrix} 10 & 20 & 30 \\ 10 & 20 & 30\end{pmatrix} = \begin{pmatrix} 11 & 22 & 33 \\ 14 & 25 & 26 \end{pmatrix} \qquad
$$

`@` でも同様のことが起こります．

````python
# B @ x
print("B @ x =")
Bxx = B @ x
print(Bxx, Bxx.shape)
print()
# B @ y
print("B @ y =")
Bxy = B @ y
print(Bxy, Bxy.shape)
````

`A @ x` に関しては，`A` の列数と `x` の要素数が 3 で等しいため，NumPy が勝手に次のような計算を行っています．

$$
\begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \end{pmatrix} \begin{pmatrix} 10 \\ 20 \\ 30 \end{pmatrix} = \begin{pmatrix} 10 + 40 + 90 \\ 40 + 100 + 180 \end{pmatrix} = \begin{pmatrix} 140 \\ 320 \end{pmatrix}
$$

`*` など他の演算子でも同様のことが起こります．

これらのことを全て理解し覚えるのは大変です．「1次元配列と2次元配列の間の演算では不思議なことが起こる」ということを頭の片隅にとどめておいて，問題が発生したら，登場する配列の shape をこまめに print して何が起こっているか確認するようにすればよいでしょう．

ちなみに，次のようにして，1次元配列を明示的に行ベクトル/列ベクトルの形の2次元配列にしてやることもできます．

````python
# 1次元配列 → 行ベクトルを表す2次元配列
x13 = x.reshape((1, 3))
print("x.reshape((1, 3)) =")
print(x13, x13.shape)
print()
# 1次元配列 → 列ベクトルを表す2次元配列
x31 = x.reshape((3, 1))
print("x.reshape((3, 1)) =")
print(x31, x31.shape)
````

### 連立方程式の解を求める

次の連立方程式を NumPy を使った数値計算で解いてみましょう．

$$
\left\{ \begin{array}{l}
2x+y = 2\\
3x-y = 8
\end{array} \right.
$$

$2\times 2$ 行列 $A$ と $2\times 1$ 行列 $\pmb{b}$ を
$$
A = \begin{pmatrix}
2 & 1\\
3 & -1
\end{pmatrix}
\qquad
\pmb{b} = \begin{pmatrix}
2 \\ 8
\end{pmatrix}
$$
とおけば，この連立方程式は
$$
A\begin{pmatrix}
x \\ y
\end{pmatrix}
= \mathbf{b}
$$
と書けます．

Numpy の関数 [`np.linalg.solve`](https://numpy.org/doc/stable/reference/generated/numpy.linalg.solve.html) に，上記の $A$ と $\pmb{b}$ に相当する np.array を 渡すと，解 $(x, y)$ の値を np.array として返してくれます．$\pmb{b}$ の方は1次元配列を渡すことに注意．

````python
A = np.array([[2, 1], [3, -1]])
b = np.array([2, 8])
x = np.linalg.solve(A, b)
print(x)
````

### ［よだんだよん］ スライスとビュー

ここは余談です．余裕のあるひと/興味のあるひと/暇なひとはどうぞ．

````python
A = np.arange(1, 13).reshape((3, 4))
print(A)
````

上記のような配列があったときに，次のようにしてその一部を参照できるのでした．

````python
print(A[1, 3])
print(A[2, :])
print(A[:, 2])
````

実は，この「`:`」の部分は，次のように書くこともできます．

````python
print(A[2, 1:3])
````

こうすると，配列 `A` の2行目の，1番目から2番目（3番目の一つ手前）までの要素を取り出すことになります． `:` の前の数を省略すると最初から，後ろの数を省略すると最後まで．

````python
print(A[2, :3])
print(A[2, 1:])
print(A[1:, :3])
````

このように `:` を使って配列の一部の要素を指定する方法を「スライス」といいます（np.array 特有のものではなく，Python のリストなどでも使えます)．

スライスを利用して，配列の一部に値を代入することもできます．

````python
A = np.arange(1, 13).reshape((3, 4))
print(A)
print()
aa = np.array([55, 66])
print(aa)
print()
A[0, 1:3] = aa
A[1:, 2] = aa
print(A)
````

スライスで切り出した配列は元の配列の一部ですので，コンピュータのメモリ上で元の配列と値を共有しています．したがって，スライスした配列の値を変更すれば，元の配列の値も変わります．

````python
A = np.arange(1, 13).reshape((3, 4))
print(A)
A[1:3, 2:] *= 10
print(A)
````

reshape 等でも同様のことになります．

````python
a = np.arange(1, 13)
A = a.reshape((3, 4))
print(a)
print(A)
print()
a[3:7] *= 10
print(a)
print(A)
````

#### 関数を使って楽をする

NumPy の関数の中に，[np.linalg.lstsq](https://numpy.org/doc/stable/reference/generated/numpy.linalg.lstsq.html) という，最小二乗法の解を求めるものがあります．これを使うと，上記の (step2) と (step3) をまとめて，1行で解を得ることができます．$X^{\top}X, X^{\top}Y$ 相当の計算を自分で書く必要がなくなります．

````python
# (step1)
xvec = np.array([1, 2, 3, 4, 5])
yvec = np.array([2, 2, -4, -6, -9])
print("xvec =", xvec)
print("yvec =", yvec)
X = np.column_stack([xvec, np.ones_like(xvec)])
print("X =")
print(X)
print(X.shape)

# (step2) & (step3)
rv = np.linalg.lstsq(X, yvec) # この関数は a, b 以外のものも返してくる
a, b = rv[0]                  # その 0 番目の要素が (a, b)
````

［よだんだよん］ lstsq の API reference に書いてありますが，戻り値は
```
( 最小二乗法の解, 残差平方和, 第1引数の行列のランク, 第1引数の行列の特異値 )
```
です．次のセルを実行すると，残差平方和の値を確認できます．

````python
print(f"残差平方和 = {rv[1][0]}")
````

ちなみに，「残差平方和」は回帰直線を用いて得られる予測値と真の値との間の差の2乗の和です．
$$
\mbox{(残差平方和)} = \sum_{n=1}^{N} (y_n - (ax_n + b))^2
$$

````python
# 自分で残差平方和を計算するならこんなん
hoge = yvec - (a*xvec+b)
print(hoge @ hoge)
````
